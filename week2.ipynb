```python
import pandas as pd

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Ahmed", "Amjad"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 4],
    "Marks": [80, 90, 75]
})

print(pd.merge(students, marks, on="ID"))                 # sirf matching (inner)
print(pd.merge(students, marks, on="ID", how="left"))     # saare students
print(pd.merge(students, marks, on="ID", how="outer"))    # dono ka sab
```

```python
# 1️⃣8️⃣ apply, map, pipe (Topic 63)
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Amjad"],
    "Marks": [50, 70, 90],
    "Gender": ["M", "F", "M"]
})

# apply -> apna function lagao
def add_bonus(m):
    return m + 5

df["New_Marks"] = df["Marks"].apply(add_bonus)

# lambda (chhota function)
df["Grade"] = df["Marks"].apply(lambda x: "Pass" if x >= 60 else "Fail")

# map -> value badalna (dictionary se)
df["Gender"] = df["Gender"].map({"M": "Male", "F": "Female"})

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Price": [100, 200, 300],
    "Quantity": [2, 3, 4],
    "Income": [50000, 60000, 70000],
    "Loan": [100000, 60000, 210000]
})

df["Total"] = df["Price"] * df["Quantity"]
df["Loan_to_Income"] = df["Loan"] / df["Income"]

# Binning -> numbers ko category banana
df["Price_Group"] = pd.cut(df["Price"], bins=[0, 150, 250, 400],
                           labels=["Low", "Medium", "High"])
print(df)
```
