```python
# 1️⃣4️⃣ Missing data (Topic 59)
df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Amjad"],
    "Age": [20, np.nan, 22],
    "Marks": [80, 75, np.nan]
})

df["Age"] = df["Age"].fillna(0)              # 0 se bharo
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())  # average se bharo

df2 = df.dropna()            # missing wali row hata do
print(df.isnull())          # True/False
print(df.isnull().sum()) 
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Amjad"],
    "Marks": [80, 95, 70]
})

print(df.sort_values("Marks"))                    # kam se zyada
print(df.sort_values("Marks", ascending=False))   # zyada se kam

df["Rank"] = df["Marks"].rank(ascending=False)
print(df)
```

```python
import pandas as pd

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Ahmed", "Amjad"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 4],
    "Marks": [80, 90, 75]
})

print(pd.merge(students, marks, on="ID"))                 # sirf matching (inner)
print(pd.merge(students, marks, on="ID", how="left"))     # saare students
print(pd.merge(students, marks, on="ID", how="outer"))    # dono ka sab
```

```python
# 1️⃣8️⃣ apply, map, pipe (Topic 63)
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Amjad"],
    "Marks": [50, 70, 90],
    "Gender": ["M", "F", "M"]
})

# apply -> apna function lagao
def add_bonus(m):
    return m + 5

df["New_Marks"] = df["Marks"].apply(add_bonus)

# lambda (chhota function)
df["Grade"] = df["Marks"].apply(lambda x: "Pass" if x >= 60 else "Fail")

# map -> value badalna (dictionary se)
df["Gender"] = df["Gender"].map({"M": "Male", "F": "Female"})

print(df)
```
